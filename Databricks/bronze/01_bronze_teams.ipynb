```python
# Location of our raw Premier League JSON file
file_path = "/Volumes/workspace/default/football_raw/premier_league_teams_20261111_225640.json"

# Read the JSON file into a Spark DataFrame
raw_df = (
    spark.read
    .option("multiLine", "true")
    .json(file_path)
)

# Inspect the structure of the data
raw_df.printSchema()

# Preview the raw dataset
display(raw_df)

# Define the destination for our Bronze Delta table
table_name = "workspace.default.bronze_teams"

# Write the raw Spark DataFrame to a managed Delta table
(
    raw_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(table_name)
)

print(f"Bronze table created: {table_name}")

# Read the saved Bronze table
bronze_df = spark.table("workspace.default.bronze_teams")

# Verify the number of records
print("Bronze rows:", bronze_df.count())

# Preview the stored data
display(bronze_df)


```
